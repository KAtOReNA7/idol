# 871 条缺图分类核对

本笔记只复算已保存证据，不执行新网络研究。代码单元由 Python 标准库 compile/exec 在仓库根目录实际运行，stdout 原样保存；不是 Jupyter 内核运行。研究结论与每条采用条件见 ledger.json。

```python
import json
from pathlib import Path
from collections import Counter
root = Path.cwd()
while not (root / "docs/codex/mv-gap-871-baseline-20261026.json").exists():
    if root == root.parent: raise RuntimeError("Run inside the idol repository")
    root = root.parent
folder = root / "docs/codex/reports/20261026-gap-remediation"
load = lambda p: json.loads(p.read_text(encoding="utf-8-sig"))
baseline = load(root / "docs/codex/mv-gap-871-baseline-20261026.json")
ledger = load(folder / "ledger.json")
audit = load(root / "docs/codex/mv-identity-asset-audit-20261025.json")
delta = load(root / "docs/codex/mv-batch-j-20261026.json")
key = lambda r: (r["workId"], r["videoId"], r["memberId"])
print({"frozen": len(baseline["rows"]), "ledger": len(ledger["entries"]), "current_missing": audit["summary"]["missingRelationshipImages"]})
```

```python
entries = ledger["entries"]
assert len(entries) == len({key(r) for r in entries}) == 871
assert {key(r) for r in entries} == {key(r) for r in baseline["rows"]}
resolved = {key(r) for r in entries if r["status"] == "resolved"}
open_rows = {key(r) for r in entries if r["status"] == "open"}
current = {(w["workId"], w["videoId"], r["memberId"]) for w in audit["workCoverage"] for r in w["missingIdentityRelationships"]}
assert resolved == {key(r) for r in delta["identityImages"]}
assert current == open_rows and len(current) == 868
assert all(r["steps"] and r["sources"] and r["acceptanceCriteria"] for r in entries)
print("PASS: exact frozen coverage, no duplicates, all action plans, 3 admitted, 868 current open tuples")
```
